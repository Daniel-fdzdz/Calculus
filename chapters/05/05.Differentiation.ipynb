## Differentiation in `Sympy`

We will do this with the `sp.diff` command, as shown in the following example:


```python
import sympy as sp
x=sp.symbols('x')
f_exp=sp.exp(x)*sp.cos(x)
d1f_exp=sp.diff(f_exp,x)
print('For the function: ',f_exp)
print('The first derivative is: ',d1f_exp)
```

## Implementing the Newton-Raphson numerical method in `Numpy`

Below we show a direct implementation of this method using a loop:


```python
import numpy as np
import sympy as sp

x = sp.symbols('x', real=True) # define the symbolic variable x

f_expr = x**3+2*x-2
f_der_expr = sp.diff(f_expr,x)

f = sp.Lambda(x,f_expr)
f_der = sp.Lambda(x,f_der_expr)

N_max = 10
tol = 1.e-9

x_aprox = np.zeros(N_max)
x_aprox[0] = 2

for k in range(1,N_max):
    if f_der(x_aprox[k-1]) == 0: break

    x_aprox[k] = x_aprox[k-1] - f(x_aprox[k-1])/f_der(x_aprox[k-1])

    if ( (k > 0) and (np.abs(x_aprox[k]-x_aprox[k-1]) / np.abs(x_aprox[k]) < tol) ): break

print('Number of iterations performed: ', k)
print('Approximation of the root: ', x_aprox[k])
```

## Successive derivatives with `Sympy`

To compute successive derivatives in `Sympy`, we must add a parameter to `sp.diff` indicating how many times we want to differentiate:


```python
import sympy as sp

x = sp.symbols('x', real=True)
f_exp = sp.sin(x) + x**2
print('Expression we want to differentiate: ',f_exp)
print('First derivative: ',sp.diff(f_exp,x))
print('Second derivative: ',sp.diff(f_exp,x,2))
print('Third derivative: ',sp.diff(f_exp,x,3))
# Note: the following notation can also be used:
# print(f_exp.diff(x,3))
```

## Using `Python` in maximum and minimum problems

Given a channel with trapezoidal cross section and side length 2, compute the angle $\alpha$ (see figure) that maximizes the area of the channel cross section.

<img src="https://github.com/maprieto/JB-Calculus1-eng-UDC/blob/main/images/cap3-canal.svg?raw=1" width="300"/>

1. By hand. Obtain the function giving the area of the channel as a function of the angle $ \alpha $
2. Symbolically: with **Sympy**.
3. Numerically using Newton's method with error smaller than $ 10^{-4} $.


```python
import sympy as sp
# 2. Solve the problem using SymPy
x,xn=sp.symbols('x,xn')
# Function describing the cross-sectional area as a function of the angle
f=4*sp.sin(x)*(1+sp.cos(x))
d1f=sp.diff(f,x)
d1fn=sp.lambdify(x,d1f)
alphamax=sp.solve(d1f)
print('The maximum cross section is attained at angle: ',float(alphamax[1]))
```

```python
# 3. Approximate the maximum with Newton's method
maxit=100
eps=1e-4
d2f=sp.diff(d1f,x)
d2fn=sp.lambdify(x,d2f)
xn=np.pi/2
for i in range(0,maxit):
    res=d1fn(xn)/d2fn(xn)
    xn=xn-res
    if (np.abs(res)<eps):
        break
print('Number of iterations performed: ',i)
print('Approximation of the angle for the maximum cross section with NR: ',xn)
```

## Computing, with `Sympy`, the Taylor polynomial and the remainder.

The following function computes symbolically the Taylor polynomial of a given function.

* Input arguments:
    * expression $f$,
    * point $x0$, which will be the Taylor center,
    * order of the polynomial, $n$.
* Output:
    * expression of the Taylor polynomial, $p$,
    * expression of the Taylor remainder, $r$.


```python
import sympy as sp

x,t=sp.symbols('x,t')

# p: Taylor polynomial
# R: remainder in absolute value
def taylor(f,x0,n):
    p=0
    for i in range(n+1):
        p+=sp.diff(f,x,i).subs(x,x0)/sp.factorial(i)*(x-x0)**i
    R=sp.diff(f,x,n+1).subs(x,t)/sp.factorial(n+1)*(x-x0)**(n+1)
    return p,R
```
